# Test perceptif v2 — le critère « note-like » est-il audible ?

**Aveugle.** Les 40 extraits sont anonymisés et mélangés. `cle.csv` contient les catégories :
**ne l'ouvre pas avant d'avoir tout noté**, sinon le test ne vaut plus rien.

## Ce qu'on cherche à savoir

La boucle d'exploration **optimise** maintenant le critère note-like : le taux est passé d'une
médiane de 0,0 % à 39,3 % sur dix graines. Mais un critère qu'on optimise cesse d'être une
mesure indépendante. Ces 39 % peuvent vouloir dire « on produit de meilleurs sons », ou
seulement « on produit des choses qui satisfont une distance de Mahalanobis sur neuf
descripteurs d'enveloppe ». Les chiffres ne peuvent pas trancher — d'où ce test.

## Barème

| note | signification |
|---|---|
| **0** | inutilisable — bruit, silence, artefact |
| **1** | passable — un son, mais pas exploitable tel quel |
| **2** | son musical utilisable |

Les extraits sont tous normalisés au même niveau crête : le volume n'est pas un indice.
Six presets humains sont cachés dans le lot pour calibrer l'échelle.

## Dépouillement

À lancer **après** avoir enregistré les notes. La clé n'est ouverte qu'ici.

```python
rep = pd.read_csv(DOSSIER / 'reponses.csv')
d = pd.read_csv(DOSSIER / 'cle.csv').merge(rep, on='id')

ORDRE = ['humain', 'box_both_notelike', 'box_both_rejete', 'box_nu', 'random']
LIB = {'humain': 'humain (controle)', 'box_both_notelike': 'box_both ACCEPTE',
       'box_both_rejete': 'box_both rejete', 'box_nu': 'temoin (sans contrainte)',
       'random': 'random'}
print(f"{'groupe':28s} {'n':>3s} {'moyenne':>8s} {'notes':>12s}")
for c in ORDRE:
    g = d[d.categorie == c]
    print(f'{LIB[c]:28s} {len(g):3d} {g.note.mean():8.2f} '
          f'{"".join(str(int(x)) for x in sorted(g.note)):>12s}')

hum = d[d.categorie == 'humain'].note.mean()
print(f"\nCONTROLE : presets humains a {hum:.2f}/2")
print('  -> protocole valide.' if hum >= 1.5 else
      '  -> ATTENTION : sous 1.5, le bareme ou la chaine de rendu est en cause, pas le critere.')

a = d[d.categorie == 'box_both_notelike'].note
r = d[d.categorie == 'box_both_rejete'].note
print(f"\nLE TEST QUI COMPTE — meme generateur, verdicts opposes :")
print(f"  acceptes {a.mean():.2f}  vs  rejetes {r.mean():.2f}   ecart {a.mean()-r.mean():+.2f}")
print('  -> le critere separe quelque chose d audible.' if a.mean() - r.mean() > 0.4 else
      '  -> le critere ne separe rien a l oreille : les 39% mesurent le critere, pas la qualite.')

n = d[d.categorie == 'box_nu'].note
print(f"\nGAIN SYSTEME : box_both accepte {a.mean():.2f} vs temoin {n.mean():.2f} "
      f"({a.mean()-n.mean():+.2f})")
```
